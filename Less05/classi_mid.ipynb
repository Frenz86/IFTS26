Esercizio 1 – Classe ContoCorrente
Traccia: Crea una classe ContoCorrente con attributi intestatario e saldo (default 0), e metodi per depositare, prelevare e mostrare il saldo.

```python
class ContoCorrente:
    def __init__(self, intestatario, saldo=0):
        # TODO: salva intestatario e saldo come attributi dell'istanza (self.intestatario, self.saldo)
        self.intestatario = intestatario
        self.saldo = saldo


    def deposita(self, importo):
        # TODO: aggiungi importo al saldo
        self.saldo = self.saldo+ importo
                # TODO: stampa un messaggio tipo: "Depositati €{importo}. Nuovo saldo: €{self.saldo}"
        print(f"Depositati €{importo}. Nuovo saldo: €{self.saldo}")

    def preleva(self, importo):
        # TODO: controlla se importo > saldo → stampa "Errore: saldo insufficiente."
        if importo>self.saldo:
            print("Errore: saldo insufficiente.")
        else:
            self.saldo= self.saldo- importo
            print(f"Prelevati €{importo}. Nuovo saldo: €{self.saldo}")
        # TODO: altrimenti sottrai importo dal saldo e stampa "Prelevati €{importo}. Nuovo saldo: €{self.saldo}"

    def mostra_saldo(self):
        print(f"Intestatario: {self.intestatario} | Saldo: €{self.saldo}")
        # TODO: stampa "Intestatario: {self.intestatario} | Saldo: €{self.saldo}"
```

```python
conto = ContoCorrente("Mario Rossi", 100)
conto.mostra_saldo()    # Intestatario: Mario Rossi | Saldo: €100
```

```python
conto.deposita(50)      # Depositati €50. Nuovo saldo: €150
```

```python
conto.preleva(200)      # Errore: saldo insufficiente.
```

```python
conto.preleva(80)       # Prelevati €80. Nuovo saldo: €70
```

Esercizio 2 – Classe Studente
Traccia: Crea una classe Studente con nome e una lista voti, con metodi per aggiungere voti, calcolare la media e verificare se è promosso.

```python
class Studente:
    def __init__(self, nome):
        # TODO: salva nome come attributo (self.nome)
        self.nome= nome
        # TODO: inizializza self.voti come lista vuota []
        self.voti=[]

    def aggiungi_voto(self, voto):
        # TODO: aggiungi voto alla lista self.voti
        self.voti.append(voto)
        print(f"Voto {voto} aggiunto per {self.nome}.")
        # TODO: stampa "Voto {voto} aggiunto per {self.nome}."

    def media(self):
        # TODO: se non ci sono voti, ritorna 0
        if len(self.voti) == 0:
            return 0
        else:
            m= sum(self.voti)/len(self.voti)
            return m

        # TODO: altrimenti ritorna la media aritmetica dei voti (sum / len)


    def promosso(self):
        # TODO: se media() >= 6 → stampa "{nome} è PROMOSSO con media {media:.1f}"
        if self.media()>=6:
            print(f"{self.nome} è PROMOSSO con media {self.media():.1f}")
        # TODO: altrimenti → stampa "{nome} è BOCCIATO con media {media:.1f}"
        else:
            print(f"{self.nome} è BOCCIATO con media {self.media():.1f}")
```

```python
s = Studente("Luca")
```

```python
s.aggiungi_voto(7)      # Voto 7 aggiunto per Luca.
```

```python
s.aggiungi_voto(5)      # Voto 5 aggiunto per Luca.
```

```python
s.aggiungi_voto(8)      # Voto 8 aggiunto per Luca.
```

```python
print(s.media())        # 6.666...
```

```python
s.promosso()            # Luca è PROMOSSO con media 6.7
```

```python
class Inventario:
    def __init__(self):
        self.prodotto={}
        # TODO: inizializza self.prodotti come dizionario vuoto {}

    def aggiungi(self, nome, quantita):
        # TODO: se nome è già nel dizionario → aggiungi quantita alla quantità esistente
        if nome not in self.prodotto.keys():
            self.prodotto[nome]=quantita
            print(f"{nome} aggiornato: {self.prodotto[nome]} pezzi.")
        else :
            self.prodotto[nome]+=quantita
            print(f"{nome} aggiornato: {self.prodotto[nome]} pezzi.")       # TODO: altrimenti → inserisci il prodotto con la quantita fornita
        # TODO: stampa "'{nome}' aggiornato: {quantita_totale} pezzi."

    def rimuovi(self, nome):
        if nome not in self.prodotto.keys():
            print(f"Errore: '{nome}' non trovato nell'inventario.")
        else:
            del self.prodotto[nome]
            print(f"'{nome}' rimosso dall'inventario.")
    #     # TODO: se nome NON è nel dizionario → stampa "Errore: '{nome}' non trovato nell'inventario."
    #     # TODO: altrimenti → rimuovi il prodotto (del) e stampa "'{nome}' rimosso dall'inventario."

    def mostra(self):
        if len(self.prodotto)==0:
            print('inventario vuoto')
        else:
            print('"=== Inventario ==="')
            for key, value in self.prodotto.items():
                print( f"{key}: {value} pezzi")

    #     # TODO: se il dizionario è vuoto → stampa "Inventario vuoto."
    #     # TODO: altrimenti stampa "=== Inventario ===" e poi ogni prodotto nel formato "  {nome}: {quantita} pezzi"
```

```python
inv = Inventario()
```

```python
inv.aggiungi("Mele", 30)        # 'Mele' aggiornato: 30 pezzi.
```

```python
inv.aggiungi("Mele", 10)        # 'Mele' aggiornato: 40 pezzi.
```

```python
inv.mostra()
# === Inventario ===
#   Mele: 40 pezzi
#   Banane: 15 pezzi
```
